**What exacty is Computer vision from Claire's POV :** is a field of artificial intelligence (AI) that enables computers and systems to derive meaningful information from digital images, videos, and other visual inputs..

###  How It Works
Computer vision relies on pattern recognition and deep learning to process visual data.
* **Data Ingestion:** The system receives visual input from devices like cameras.
* **Feature Extraction:** Algorithms break down the image into pixel data to identify edges, colors, and textures.
* **Deep Learning Models:** Convolutional Neural Networks (CNNs) analyze these features by comparing them to millions of pre-trained training images.
* **Actionable Output:** The system interprets the image and delivers a conclusion, such as identifying a defect or triggering an alarm.

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

img = mpimg.imread('clae.jpg')
r = img[:, :, 0]
g = img[:, :, 1]
b = img[:, :, 2]

plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(gray_image, cmap = 'gray')
```

```python
r = img[:,:,0]
g = img[:,:,1]
b = img[:,:,2]

f, (ax1, ax2, ax3) = plt.subplots(1,3,figsize = (20, 10))
ax1.set_title('R channel')
ax1.imshow(r, cmap= 'gray')
ax2.set_title('G channel')
ax2.imshow(r, cmap= 'gray')
ax3.set_title('B channel')
ax3.imshow(r, cmap= 'gray')
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
h = hsv [:,:,0]
s = hsv[:,:,1]
v = hsv[:,:,2]

f, (ax1, ax2, ax3) = plt.subplots(1,3,figsize = (20, 10))
ax1.set_title('Hue')
ax1.imshow(h, cmap = 'gray')
ax2.set_title('Saturation')
ax2.imshow(s, cmap = 'gray')
ax3.set_title('Value')
ax3.imshow(v, cmap = 'gray')
```
